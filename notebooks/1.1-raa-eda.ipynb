```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing import image
import gradio as gr
```

```python
TARGET_SIZE = (600, 600)

model = tf.keras.models.load_model("animal_detection_model.h5")

class_names_espanol = [
    "Oso", "Oso pardo", "Toro", "Mariposa", "Camello", "Canario", "Oruga", "Ganado", "Ciempiés", "Guepardo",
    "Pollo", "Cangrejo", "Cocodrilo", "Ciervo", "Pato", "Águila", "Elefante", "Pez", "Zorro", "Rana",
    "Jirafa", "Cabra", "Pez dorado", "Ganso", "Hámster", "Foca común", "Erizo", "Hipopótamo", "Caballo", "Jaguar",
    "Medusa", "Canguro", "Koala", "Mariquita", "Leopardo", "León", "Lagarto", "Lince", "Urraca", "Mono",
    "Polillas y mariposas", "Ratón", "Mula", "Avestruz", "Nutria", "Búho", "Panda", "Loro", "Pingüino", "Cerdo",
    "Oso polar", "Conejo", "Mapache", "Cuervo", "Panda rojo", "Rinoceronte", "Escorpión", "León marino", "Tortuga marina", "Caballito de mar",
    "Tiburón", "Oveja", "Camarón", "Caracol", "Serpiente", "Gorrión", "Araña", "Calamar", "Ardilla", "Estrella de mar",
    "Cisne", "Garrapata", "Tigre", "Tortuga terrestre", "Pavo", "Tortuga", "Ballena", "Pájaro carpintero", "Gusano", "Cebra"
]
```

```python
def predict_image_gradio(img):
    img = img.convert("RGB").resize(TARGET_SIZE)
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis=0)
    img_array = tf.keras.applications.efficientnet.preprocess_input(img_array)

    predictions = model.predict(img_array)[0]

    if predictions.max() > 1.0:
        predictions = tf.nn.softmax(predictions).numpy()

    top_5_indices = np.argsort(predictions)[::-1][:5]
    resultados = {
        f"{i:02d} - {class_names_espanol[i]}": round(float(predictions[i]), 4)
        for i in top_5_indices
    }

    return resultados
```

```python
iface = gr.Interface(
    fn=predict_image_gradio,
    inputs=gr.Image(type="pil"),
    outputs=gr.Label(num_top_classes=5),
    title="🦁 Clasificador de Imágenes de Animales",
    description="Sube una imagen de un animal para ver las 5 predicciones principales del modelo EfficientNetB7 entrenado.",
    allow_flagging="never"
)

iface.launch(share=True)
```
